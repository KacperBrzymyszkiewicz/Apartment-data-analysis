```python
import os
import pandas as pd
import plotly.express as pt
data = []
cols_num = 0
for file in os.listdir("./data"):
    if file.endswith(".csv") and "rent" not in file.lower():
        df = pd.read_csv(os.path.join("./data", file))
        if cols_num == 0:
            cols_num = df.shape[1]
            data.append(df)
        elif cols_num == df.shape[1]:
            data.append(df)
        else:
            print(f"{file} has a different number of columns")
data = pd.concat(data,ignore_index=True)
data["priceForSquareMeter"] = round(data["price"]/data["squareMeters"],2)
#print(data.columns.tolist())
#print(data[data["price"]<= 100000].count())
print(data[['price','squareMeters','type']].sort_values('price',ascending=True).head(50))
#print(data["city"].value_counts())
# print(data["priceForSquareMeter"])
# print(data["price"].describe())
```

```python
sqmeterplot = pt.scatter(data[data["city"]=="poznan"],x="priceForSquareMeter",y="squareMeters",color="city",size="squareMeters")
sqmeterplot.show()
```

```python
cityplot = pt.histogram(data,x="city")
cityplot.show()
```

```python
cityPricePlot = pt.histogram(data,x="city",y="priceForSquareMeter",histfunc="avg",color="city").update_xaxes(categoryorder="total descending")
cityPricePlot.show()
```
